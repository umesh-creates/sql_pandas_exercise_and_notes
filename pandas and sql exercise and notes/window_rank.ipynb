**RANK()**

<pre>Har state ke products ko product price ke hisaab se sabse mehange se sabse saste tak rank do.
Agar do products ki price same ho, to dono ko same rank milna chahiye.</pre>

```python
%%sql
SELECT order_state, product_name, product_price,
RANK()
OVER(
    PARTITION BY order_state
    ORDER BY product_price DESC
) AS product_price_rank_bottom_top
FROM products01;
```

```python
products = (
    products
    .sort_values(
        ['order_state', 'product_price'],
        ascending=[True, False]
    )
)
products['product_price_rank_bottom_top'] = (
    products
    .groupby('order_state')['product_price']
    .rank(
        ascending=False,
        method = 'min'
    )
)
products[
    ['order_state', 'product_name', 'product_price', 'product_price_rank_bottom_top']
]
```

**Method = 'average' (default)**
<pre>Value	Rank
50000	1.5
50000	1.5
2000	3.0</pre>
<pre>2 times 50000, so (1+2)/2 = 1.5 rank</pre>

**method='min' ⭐ (SQL RANK())**
<pre>Value	Rank
50000	1
50000	1
2000	3</pre>

**method='dense' ⭐ (SQL DENSE_RANK())**
<pre>Value	Rank
50000	1
50000	1
2000	2</pre>

**method='first'** \
like **cumcount** of pandas and **row_number** of sql
<pre>Value	Rank
50000	1
50000	2
2000	3</pre>

<pre>Har customer ke products ko sabse zyada discount se sabse kam discount tak rank do.
Agar do orders ka discount same ho, to dono ko same rank milna chahiye aur agla rank skip hona chahiye.

SQL Output Columns
customer_name

discount_rank_high_to_low</pre>

```python
%%sql 
SELECT customer_name, product_name, discount, 
RANK()
OVER(
    PARTITION BY customer_name
    ORDER BY discount DESC
) AS discount_rank_high_to_low
FROM products01;
```

```python
products = (
    products
    .sort_values(
        ['customer_name', 'discount'],
        ascending=[True, False]
    )
)
products['discount_rank_high_to_low'] = (
    products
    .groupby('customer_name')['discount']
    .rank(
        ascending=False,
        method='min'
    )
)
products[
    ['customer_name', 'product_name', 'discount', 'discount_rank_high_to_low']
]
```

<pre>Har state ke products ko sabse mehange se sabse saste tak RANK() do.
Agar do products ki price same ho, to same rank mile.
Phir sirf Top 2 ranks wale products hi dikhao.</pre>

```python
%%sql
WITH ranked_order AS (
    SELECT order_state, product_name, product_price,
    RANK()
    OVER(
        PARTITION BY order_state
        ORDER BY product_price DESC
    ) AS product_price_high_to_low
    FROM products01
)
SELECT order_state, product_name, product_price, product_price_high_to_low
FROM ranked_order
WHERE product_price_high_to_low <= 2;
```

```python
products = (
    products
    .sort_values(
        ['order_state', 'product_price'],
        ascending=[True, False]
    )
)
products['product_price_high_to_low'] = (
    products
    .groupby('order_state')['product_price']
    .rank(
        ascending=False,
        method='min'
    )
)
products[
    ['order_state', 'product_name', 'product_price', 'product_price_high_to_low']
][products['product_price_high_to_low'] <= 2]
```

```python
products = (
    products
    .sort_values(
        ['order_state', 'product_price', 'order_date'],
        ascending=[True, False, True]
    )
)
products['price_rank_if_same_old_order_first'] = (
    products
    .groupby('order_state')[['product_price','order_date']]
    .rank(
        ascending=[False, True],
        method='min'
    )
)
products[
    ['order_state', 'product_name', 'product_price', 'order_date', 'price_rank_if_same_old_order_first']
]
```
