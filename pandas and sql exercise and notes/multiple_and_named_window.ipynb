### **multiple and named window**

<pre>Har **order_state** ke andar **net_amount** DESC ke hisaab se:

ROW_NUMBER()
RANK()
DENSE_RANK()</pre>

```python
%%sql 
SELECT order_state, net_amount,
ROW_NUMBER()
OVER(
    PARTITION BY order_state
    ORDER BY net_amount DESC
) AS row_number,

RANK()
OVER(
    PARTITION BY order_state
    ORDER BY net_amount DESC
) AS rank,

DENSE_RANK()
OVER(
    PARTITION BY order_state
    ORDER BY net_amount DESC
) AS dense_rank
FROM products01;
```

```python
%%sql 
SELECT order_state, net_amount,
ROW_NUMBER()
OVER w AS row_number,

RANK()
OVER w AS rank,

DENSE_RANK()
OVER w AS dense_rank

FROM products01

WINDOW w AS (
    PARTITION BY order_state
    ORDER BY net_amount DESC
);
```
