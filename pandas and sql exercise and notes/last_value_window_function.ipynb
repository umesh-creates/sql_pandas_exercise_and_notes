### **LAST_VALUE()**
**Returns the last value in the current window.**

<PRE>Window Function =

1. PARTITION BY
2. ORDER BY
3. WINDOW FRAME</PRE>

<pre>Table ke **order_date** ke adhar par last **net_amount** dikhao.</pre>

```python
%%sql 
SELECT order_date, net_amount,
LAST_VALUE(net_amount)
OVER(
    ORDER BY order_date ASC
    ROWS BETWEEN UNBOUNDED PRECEDING
    AND UNBOUNDED FOLLOWING
) AS last_net_amount 
FROM products01;
```

```python
products = (
    products
    .sort_values('order_date')
)
products['last_net_amount'] = (
    products['net_amount']
    .iloc[-1]
)
products[
    ['order_date', 'net_amount', 'last_net_amount']
]
```

<pre>Har **customer_name** ke andar **order_date** ke adhar par last **net_amount** dikhao.</pre>

```python
%%sql 
SELECT customer_name, order_date, net_amount,
LAST_VALUE(net_amount)
OVER(
    PARTITION BY customer_name
    ORDER BY order_date ASC
    ROWS BETWEEN UNBOUNDED PRECEDING
    AND UNBOUNDED FOLLOWING
) AS last_net_amount 
FROM products01;
```

```python
products = (
    products
    .sort_values(
        ['customer_name', 'order_date'],
        ascending=[True, True]
    )
)
products['last_net_amount'] = (
    products
    .groupby('customer_name')['net_amount']
    .transform('last')
)
products[ 
    ['customer_name', 'order_date', 'net_amount', 'last_net_amount']
]
```
