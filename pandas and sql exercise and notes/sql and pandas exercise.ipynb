```python
%%sql 
CREATE TABLE products(
    product_id SERIAL PRIMARY KEY ,
    product_name VARCHAR(50) NOT NULL,
    order_date DATE NOT NULL,
    customer_name VARCHAR(50) NOT NULL,
    order_state VARCHAR(50) NOT NULL,

    product_price NUMERIC(10,2) NOT NULL
        CHECK(product_price >= 0),

    product_quantity INT NOT NULL
        CHECK (product_quantity > 0),
    
    discount NUMERIC(5,2) DEFAULT 0 
        CHECK (discount >= 0 AND discount <= 50),

    gross_amount NUMERIC(10,2)
        GENERATED ALWAYS AS
        (product_price * product_quantity) 
        STORED,

    net_amount NUMERIC(10,2)
        GENERATED ALWAYS AS
        ((product_price * product_quantity) * (1 - discount / 100)) 
        STORED
);
```

```python
%%sql 
SELECT product_name, customer_name, discount, MAX(discount) 
OVER(PARTITION BY product_name) AS product_max_discount
FROM products;
```

```python
products['product_max_discount'] = (
    products
    .groupby('product_name')['discount']
    .transform('max')
)
products[
    ['product_name', 'customer_name', 'discount', 'product_max_discount']
]
```

<pre>Har order ke saath us state me sabse kam (minimum) product price bhi dikhani hai.
customer_name	order_state	product_price	state_min_price</pre>

```python
%%sql 
SELECT customer_name, order_state, product_price, MIN(product_price)
OVER (PARTITION BY order_state) AS state_min_price
FROM products;
```

```python
products['state_min_price'] = (
    products
    .groupby('order_state')['product_price']
    .transform('min')
)
products[
    ['customer_name', 'order_state', 'product_price', 'state_min_price']
]
```

<pre>Har order ke saath us state ka:

1. Total Revenue
2. Maximum Product Price
customer_name   order_state   net_amount   product_price   state_total_revenue   state_max_price</pre>

```python
%%sql 
SELECT customer_name, order_state, net_amount, product_price, SUM(net_amount)
OVER(PARTITION BY order_state) AS state_total_revenue,
MAX(product_price)
OVER(PARTITION BY order_state) AS state_max_price
FROM products;
```

```python
products['state_total_revenue'] = (
    products
    .groupby('order_state')['net_amount']
    .transform('sum')
)
products['state_max_price'] = (
    products
    .groupby('order_state')['product_price']
    .transform('max')
)
products[
    ['customer_name', 'order_state', 'net_amount', 'product_price', 'state_total_revenue', 'state_max_price']
]
```
