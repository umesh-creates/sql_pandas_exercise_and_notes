### **CUME_DIST()**
**Returns the cumulative distribution of a value within a group.**

<pre>100 -> 0.25 -> 1/4
90  -> 0.75 -> 3/4
90  -> 0.75 -> 3/4
80  -> 1.00 -> 4/4</pre>

<pre>Whole table me **discount** ke according highest se lowest tak CUME_DIST() nikalo.</pre>

```python
%%sql 
SELECT discount,
CUME_DIST()
OVER(
    ORDER BY discount DESC
) AS cumulative_distribution
FROM products01;
```

<pre>Har **order_state** ke andar **discount** ke according highest se lowest tak CUME_DIST() nikalo.</pre>

```python
%%sql 
SELECT order_state, discount,
CUME_DIST()
OVER(
    PARTITION BY order_state
    ORDER BY discount DESC
) AS cumulative_distribution 
FROM products01;
```
