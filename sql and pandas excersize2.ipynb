```python
%%sql 
CREATE TABLE products(
    product_id SERIAL PRIMARY KEY ,
    product_name VARCHAR(50) NOT NULL,
    order_date DATE NOT NULL,
    customer_name VARCHAR(50) NOT NULL,
    order_state VARCHAR(50) NOT NULL,

    product_price NUMERIC(10,2) NOT NULL
        CHECK(product_price >= 0),

    product_quantity INT NOT NULL
        CHECK (product_quantity > 0),
    
    discount NUMERIC(5,2) DEFAULT 0 
        CHECK (discount >= 0 AND discount <= 50),

    gross_amount NUMERIC(10,2)
        GENERATED ALWAYS AS
        (product_price * product_quantity) 
        STORED,

    net_amount NUMERIC(10,2)
        GENERATED ALWAYS AS
        ((product_price * product_quantity) * (1 - discount / 100)) 
        STORED
);
```

**WINDOW ROW_NUMBER() FUNCTION**

<pre>Har state ke andar products ko <i>product_price</i> ke hisaab se sabse mehange se sabse saste tak rank nahi, balki ek serial number dena hai.

Expected Output (Example):

order_state  	product_name	product_price	row_number</pre>

```python
%%sql 
SELECT order_state, product_name, product_price, 
ROW_NUMBER()
OVER(
    PARTITION BY order_state
    ORDER BY product_price DESC
) AS row_number
FROM products;
```

```python
products = (
    products
    .sort_values(
        ['order_state', 'product_price'],
        ascending = [True, False]
    )
)
products['row_number'] = (
    products
    .groupby('order_state')
    .cumcount()
    + 1
)
products[
    ['order_state', 'product_name', 'product_price', 'row_number']
]
```

<pre>Har customer ke orders ko order_date ke hisaab se purane se naye tak serial number do.

Expected Output (Example)

customer_name	order_date	product_name	order_number</pre>

```python
%%sql
SELECT customer_name, order_date, product_name,
ROW_NUMBER()
OVER(
    PARTITION BY customer_name
    ORDER BY order_date ASC
) AS order_number
FROM products;
```

```python
products = (
    products
    .sort_values(
        ['customer_name', 'order_date'],
        ascending = [True, True]
    )
)
products['order_number'] = (
    products
    .groupby('customer_name')
    .cumcount()
    + 1
)
products[
    ['customer_name', 'order_date', 'product_price', 'order_number']
]
```

<pre>Har product ke liye sabse recent order ko 1 number milna chahiye.</pre>

```python
%%sql 
WITH ranked_orders AS (
    SELECT product_name, order_date,
    ROW_NUMBER()
    OVER(
        PARTITION BY product_name
        ORDER BY order_date DESC
    ) AS recent_order
    FROM products
)
SELECT product_name, order_date, recent_order
FROM ranked_orders
WHERE recent_order < 2;
```

```python
products = (
    products
    .sort_values(
        ['product_name', 'order_date'],
        ascending = [True, False]
    )
)
products['recent_order'] = (
    products
    .groupby('product_name')
    .cumcount()
    + 1 
)
products.loc[
    products['recent_order'] == 1,
    ['product_name', 'order_date', 'recent_order']
]        
```

<pre>Har state ke orders ko pehle product price (high to low) aur agar price same ho to order date (old to new) ke hisaab se serial number do.</pre>

```python
%%sql 
SELECT order_state, product_price,
ROW_NUMBER()
OVER(
    PARTITION BY order_state
    ORDER BY product_price DESC, order_date ASC
) AS product_price_high_low
FROM products;
```

```python
products = (
    products
    .sort_values(
        ['order_state', 'product_price', 'order_date'],
        ascending=[True, False, True]
    )
)
products['price_high_to_low_if_same_date_old_to_new'] = (
    products
    .groupby('order_state')
    .cumcount()
    + 1
)
products[
    ['order_state', 'product_price', 'price_high_to_low_if_same_date_old_to_new']
]
```

```python
products = (
    products
    .sort_values(
        ['customer_name', 'product_quantity', 'discount'],
        ascending = (True, False, False)
    )
)
products['quantity_high_to_low_if_same_discount_high_to_low'] = (
    products
    .groupby('customer_name')
    .cumcount()
    + 1
)
products[
    ['customer_name', 'product_quantity', 'quantity_high_to_low_if_same_discount_high_to_low']
]
```
