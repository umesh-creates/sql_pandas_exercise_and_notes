Next Exercise — Monthly Revenue

Using orders, order_items, and products, calculate the total revenue for each month.

Return:

order_month \
total_revenue

Use:

selling_price × quantity

and DATE_TRUNC('month', order_date) for the month.

```python
%%sql 
SELECT DATE_TRUNC('month', order_date) AS order_month,
    SUM(oi.quantity * p.selling_price) AS total_revenue
FROM retail_analysis.orders o 
JOIN retail_analysis.order_items oi 
ON o.order_id = oi.order_id
JOIN retail_analysis.products p 
ON oi.product_id = p.product_id
GROUP BY DATE_TRUNC('month', order_date)
ORDER BY order_month ASC;
```

Next Exercise — Monthly Average Order Value

Using orders, order_items, and products, calculate **the average order value for each month**.

**Important**: first **calculate the revenue of each individual order**, then **calculate the average of those order revenues for each month**.

```python
%%sql 
SELECT order_month, AVG(revenue) AS avg_order_value
FROM (
    SELECT o.order_id, 
        DATE_TRUNC('month', o.order_date) AS order_month,
        SUM(oi.quantity * p.selling_price) AS revenue
    FROM retail_analysis.orders o 
    JOIN retail_analysis.order_items oi 
    ON oi.order_id = o.order_id 
    JOIN retail_analysis.products p
    ON p.product_id = oi.product_id
    GROUP BY o.order_id, DATE_TRUNC('month', o.order_date)
) 
GROUP BY order_month;
```

Next Exercise — Monthly Revenue Growth

Using orders, order_items, and products, calculate the **total revenue for each month and also show the previous month's revenue alongside it**.

Return:

order_month \
total_revenue \
previous_month_revenue

You may use a window function such as **LAG()**.

```python
%%sql
SELECT DATE_TRUNC('month', o.order_date) AS month,
    SUM(p.selling_price * oi.quantity) total_revenue,
    LAG(SUM(p.selling_price * oi.quantity))
    OVER(
        ORDER BY DATE_TRUNC('month', o.order_date) ASC
    ) AS previous_month_revenue
FROM retail_analysis.orders o 
JOIN retail_analysis.order_items oi 
ON o.order_id = oi.order_id
JOIN retail_analysis.products p 
ON p.product_id = oi.product_id
GROUP BY DATE_TRUNC('month', o.order_date);
```

Next Exercise — Monthly Revenue Change

Using the same monthly revenue logic, calculate:

order_month \
total_revenue \
previous_month_revenue \
revenue_change

Where:

revenue_change = total_revenue - previous_month_revenue

Use LAG().

```python
%%sql 
SELECT order_month, 
    total_revenue, 
    previous_month_revenue, 
    (total_revenue - previous_month_revenue) AS revenue_change
FROM (
    SELECT DATE_TRUNC('month', o.order_date) AS order_month, 
        SUM(p.selling_price * oi.quantity) AS total_revenue, 
        LAG(SUM(p.selling_price * oi.quantity))
        OVER(
            ORDER BY DATE_TRUNC('month', o.order_date) ASC
        ) AS previous_month_revenue
    FROM retail_analysis.orders o 
    JOIN retail_analysis.order_items oi 
    ON oi.order_id = o.order_id
    JOIN retail_analysis.products p
    ON p.product_id = oi.product_id
    GROUP BY DATE_TRUNC('month', o.order_date)
)
```

Next Exercise — Revenue Growth %

Now calculate:

order_month \
total_revenue \
previous_month_revenue \
revenue_change \
revenue_growth_pct

Formula:

revenue_growth_pct = \
(revenue_change / previous_month_revenue) × 100

Use NULLIF() so you don't get a division-by-zero error.

```python
%%sql 
SELECT order_month, total_revenue, previous_month_revenue, 
    (total_revenue - previous_month_revenue) AS revenue_change,
    (
        (total_revenue - previous_month_revenue)
        / NULLIF(previous_month_revenue, 0)
    ) * 100 AS revenue_growth_pct
FROM (
    SELECT DATE_TRUNC('month', o.order_date) AS order_month,
        SUM(p.selling_price * oi.quantity) AS total_revenue,
        LAG(SUM(p.selling_price * oi.quantity))
        OVER(
            ORDER BY DATE_TRUNC('month', o.order_date) ASC
        ) AS previous_month_revenue
    FROM retail_analysis.orders o
    JOIN retail_analysis.order_items oi 
    ON oi.order_id = o.order_id 
    JOIN retail_analysis.products p 
    ON p.product_id = oi.product_id
    GROUP BY DATE_TRUNC('month', o.order_date)
    ORDER BY order_month ASC
)
```
