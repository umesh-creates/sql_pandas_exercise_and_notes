Next Exercise — Customer Ordering Behavior

For each customer, calculate the number of days between their first and last order.

Return:

customer_id \
first_order_date \
last_order_date \
days_between_orders 

Use MIN(), MAX(), and a PostgreSQL date/time function to calculate the difference in total days.

```python
%%sql 
SELECT customer_id, 
    MIN(order_date) AS first_order_date,
    MAX(order_date) AS last_order_date,
    EXTRACT(EPOCH FROM AGE(MAX(order_date), MIN(order_date))) / 86400 AS days_between_orders
FROM retail_analysis.orders
GROUP BY customer_id;
```

```python
result = (
    orders
    .groupby('customer_id')
    .agg(
        first_order_date = ('order_date', 'min'),
        last_order_date = ('order_date', 'max')
    )
    .reset_index()
)
result['days_between_orders'] = (
    result['last_order_date'].dt.day - result['first_order_date'].dt.day
)
result
```

Next Exercise — Monthly Revenue Growth

You may use a window function such as **LAG()**.

```python
%%sql
SELECT DATE_TRUNC('month', o.order_date) AS month,
    SUM(p.selling_price * oi.quantity) total_revenue,
    LAG(SUM(p.selling_price * oi.quantity))
    OVER(
        ORDER BY DATE_TRUNC('month', o.order_date) ASC
    ) AS previous_month_revenue
FROM retail_analysis.orders o 
JOIN retail_analysis.order_items oi 
ON o.order_id = oi.order_id
JOIN retail_analysis.products p 
ON p.product_id = oi.product_id
GROUP BY DATE_TRUNC('month', o.order_date);
```

Next Exercise — Revenue Growth %

Now calculate:

Formula:

revenue_growth_pct = \
(revenue_change / previous_month_revenue) × 100

Use NULLIF() so you don't get a division-by-zero error.

```python
%%sql 
SELECT order_month, total_revenue, previous_month_revenue, 
    (total_revenue - previous_month_revenue) AS revenue_change,
    (
        (total_revenue - previous_month_revenue)
        / NULLIF(previous_month_revenue, 0)
    ) * 100 AS revenue_growth_pct
FROM (
    SELECT DATE_TRUNC('month', o.order_date) AS order_month,
        SUM(p.selling_price * oi.quantity) AS total_revenue,
        LAG(SUM(p.selling_price * oi.quantity))
        OVER(
            ORDER BY DATE_TRUNC('month', o.order_date) ASC
        ) AS previous_month_revenue
    FROM retail_analysis.orders o
    JOIN retail_analysis.order_items oi 
    ON oi.order_id = o.order_id 
    JOIN retail_analysis.products p 
    ON p.product_id = oi.product_id
    GROUP BY DATE_TRUNC('month', o.order_date)
    ORDER BY order_month ASC
)
```

```python
merged = (
    orders
    .merge(order_items, on='order_id')
    .merge(products, on='product_id')
)
result = (
    merged 
    .assign(revenue = merged['selling_price'] * merged['quantity'])
    .groupby(merged['order_date'].dt.to_period('M'))['revenue']
    .sum()
    .reset_index(name='total_revenue')
    .rename(columns={'order_date' : 'order_month'})
)
result['previous_month_revenue'] = (
    result['total_revenue']
    .shift(1)
)
result['revenue_change'] = (
    result['total_revenue'] - result['previous_month_revenue']
)
result['revenue_growth_pct'] = (
    (result['revenue_change'] /
    result['previous_month_revenue'].mask(
        result['previous_month_revenue'] == 0)
    )
    * 100
)
result
```

Next Exercise — Order Gap

For **each customer**, find the **number of days between their first and second order.**

Return:

customer_id \
first_order_date \
second_order_date \
days_between

Use ROW_NUMBER(), AGE(), and EPOCH.

```python
%%sql 
SELECT customer_id, first_order_date, second_order_date,
    EXTRACT(EPOCH FROM AGE(second_order_date, first_order_date)) / 86400 AS days_between
FROM (
    SELECT customer_id, 
        MIN(order_date) AS first_order_date,
        MAX(order_date) AS second_order_date
    FROM (
        SELECT customer_id, order_date,
        ROW_NUMBER()
        OVER(
            PARTITION BY customer_id
            ORDER BY order_date ASC
        ) AS order_number
        FROM retail_analysis.orders
    )
    WHERE order_number IN (1,2)
    GROUP BY customer_id
)
```

Next Exercise — Customer Purchase Frequency

For **each customer**, **calculate their average number of days between orders**.

Return:

customer_id \
average_days_between_orders

Hint: first calculate the gap between consecutive orders, then take AVG() for each customer.

Use LAG(), AGE(), and EPOCH().

```python
%%sql 
SELECT customer_id, AVG(days_between_orders) AS avg_days_between_orders
FROM (
    SELECT customer_id,
        EXTRACT(EPOCH FROM AGE(order_date, previous_date)) / 86400 AS days_between_orders
    FROM (
        SELECT customer_id, order_date,
        LAG(order_date)
        OVER(
            PARTITION BY customer_id
            ORDER BY order_date ASC
        ) AS previous_date
        FROM retail_analysis.orders
    )
)
GROUP BY customer_id
```

For **each month**, **count the customers who placed at least 2 orders during that month**.

repeat_customers

```python
%%sql
SELECT order_month,
    COUNT(customer_id) AS repeat_customers
FROM (
    SELECT customer_id,
        DATE_TRUNC('month', order_date) AS order_month,
        COUNT(order_id) AS total_orders
    FROM retail_analysis.orders
    GROUP BY customer_id, DATE_TRUNC('month', order_date)
    HAVING COUNT(order_id) >= 2
) AS monthly_customer_orders
GROUP BY order_month
ORDER BY order_month;
```
