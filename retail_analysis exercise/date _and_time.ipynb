### **DATE TIME**

```python
%%sql 
CREATE TABLE retail_analysis.customers(
    customer_id VARCHAR(20) PRIMARY KEY,
    customer_name VARCHAR(50) NOT NULL,
    gender VARCHAR(20) NOT NULL,
    birth_date DATE NOT NULL,
    city VARCHAR(50) NOT NULL,
    state VARCHAR(50) NOT NULL, 
    join_date DATE NOT NULL
)
```

```python
%%sql 
CREATE TABLE retail_analysis.orders (
    order_id VARCHAR(20) PRIMARY KEY,
    customer_id VARCHAR(20) NOT NULL,
    login_date TIMESTAMP NOT NULL,
    order_status VARCHAR(20) NOT NULL,
    order_date TIMESTAMP NOT NULL,
    logout_date TIMESTAMP NOT NULL,
    delivery_date TIMESTAMP,
    payment_mode VARCHAR(20),
    time_zone VARCHAR(50) NOT NULL,

    FOREIGN KEY (customer_id)
    REFERENCES retail_analysis.customers(customer_id)
)
```

Date & Time Analysis — Question 1

Find the `customer_name` and their **age in years** as of today using `birth_date`. Use PostgreSQL date/time functions.

```python
%%sql 
SELECT customer_name, EXTRACT(YEAR FROM AGE(birth_date)) AS age
FROM retail_analysis.customers;
```

Next Question

Find each customer's `customer_name` and the number of complete years they have been a customer, calculated from `join_date` up to today. Use PostgreSQL date/time functions.

```python
%%sql 
SELECT customer_name, EXTRACT(YEAR FROM AGE(join_date)) AS years_of_customer
FROM retail_analysis.customers;
```

Next Question

For **each order**, display `order_id`, `order_date`, and the **year, month, and day** of the **order** separately. Use EXTRACT().

```python
%%sql 
SELECT order_id, order_date,
    EXTRACT(YEAR FROM order_date) AS year,
    EXTRACT(MONTH FROM order_date) AS month,
    EXTRACT(DAY FROM order_date) AS day
FROM retail_analysis.orders;
```

Next Question

For **each order**, display `order_id`, `order_date`, and the **hour, minute, and second** when the order was placed. Use EXTRACT().

```python
%%sql 
SELECT order_id, order_date,
    EXTRACT(HOUR FROM order_date) AS hours,
    EXTRACT(MINUTE FROM order_date) AS minute,
    EXTRACT(SECOND FROM order_date) AS second
FROM retail_analysis.orders;
```

Next Question

For **each order**, calculate **how much time passed between order_date and delivery_date**. Display `order_id`, `order_date`, `delivery_date`, and the `delivery duration`. Use PostgreSQL date/time functions.

```python
%%sql 
SELECT order_id, order_date, delivery_date,
    AGE(delivery_date, order_date) AS delivery_duration
FROM retail_analysis.orders;
```

Next Question

For **each order**, calculate the **total delivery time in hours** between `order_date` and `delivery_date`. Display `order_id` and **delivery_hours**. Use PostgreSQL date/time functions.

```python
%%sql 
SELECT order_date, delivery_date, order_id,
    EXTRACT(EPOCH FROM AGE(delivery_date, order_date)) / 3600 AS delivery_hours
FROM retail_analysis.orders;
```

Next Question

For **each order**, calculate the **total delivery time in days**, **including fractional days**. Display `order_id` and `delivery_days`. Use PostgreSQL date/time functions.

```python
%%sql 
SELECT order_id, delivery_date, order_date,
    EXTRACT(EPOCH FROM AGE(delivery_date, order_date)) / 210000 AS delivery_days
FROM retail_analysis.orders;
```
