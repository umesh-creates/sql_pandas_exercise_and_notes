From the orders table, find the **total number of orders** and **the number of orders that have the status 'Delivered'.** Return both values in a single row.

```python
%%sql 
SELECT COUNT(order_id) AS total_orders,
    COUNT(order_id) FILTER (WHERE order_status = 'completed') AS delivered_orders
FROM retail_analysis.orders
```

Question:

From the products table, display `product_id`, `cost_price`, and `selling_price`, and create a new column called `higher_price` that contains the **higher value between `cost_price` and `selling_price`**.

```python
%%sql
SELECT product_id, cost_price, selling_price,
    GREATEST(cost_price, selling_price) AS highest_price
FROM retail_analysis.products
```

axis=0 → नीचे की तरफ → rows को aggregate करो \
axis=1 → बगल में → columns को aggregate करो

```python

result = products[
    ['product_id', 'cost_price', 'selling_price']
]
result['highest_price'] = result[['cost_price', 'selling_price']].max(axis=1)
result.head()
```

Exercise — GREATEST()

From the products table, display `product_id`, `cost_price`, and `selling_price`. Create a new column called `maximum_price` that **contains the higher price between** `cost_price` and `selling_price`.

```python
%%sql 
SELECT product_id, cost_price, selling_price,
    GREATEST(cost_price, selling_price) AS maximum_price
FROM retail_analysis.products
```

```python
result = products[
    ['product_id', 'cost_price', 'selling_price']
]
result['maximum_price'] = result[['cost_price', 'selling_price']].max(axis=1)
result.head()
```

### **LEAST()**

Exercise — LEAST()

From the products table, display `product_id`, `cost_price`, and `selling_price`. Create a new column called `lowest_price` that **contains the lower value between `cost_price` and `selling_price`.**

```python
%%sql 
SELECT product_id, cost_price, selling_price, 
    LEAST(cost_price, selling_price) AS lowest_price
FROM retail_analysis.products
```

```python
result = products[
    ['product_id', 'cost_price', 'selling_price']
]
result['lowest_price'] = result[['cost_price', 'selling_price']].min(axis=1)
result.head()
```

Question:

From the products table, calculate the total `selling_price` for each `category` and `brand`. Also **include a subtotal for each category** and a **grand total for all products**. Use ROLLUP.

```python
%%sql 
SELECT category, brand, SUM(selling_price) AS total_selling_price
FROM retail_analysis.products
GROUP BY ROLLUP(category, brand)
```

```python
%%sql
SELECT
    COALESCE(category, 'Grand Total') AS category,
    COALESCE(brand, 'Total') AS brand,
    SUM(selling_price) AS total_selling_price
FROM retail_analysis.products
GROUP BY ROLLUP(category, brand);
```

```python
%%sql 
SELECT category, brand, SUM(selling_price) AS total_selling_price
FROM retail_analysis.products
GROUP BY CUBE(category, brand)
```
