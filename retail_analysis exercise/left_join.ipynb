<pre>Question 27 — JOIN + Aggregation

Find the total number of products purchased by each customer. 
Include customers who have never placed an order. 
Display customer_id, customer_name, and total_products_purchased.</pre>

<pre>
DISTINCT customers[customer_id],
DISTINCT customers[customer_name],
order_items[quantity].sum() as total_product_purchased
customers 1* 2*
order_items 1* 3
LEFT JOIN
</pre>

**COALESCE(null, 23) \
NULL values replace with , num**

```python
merged = (
    customers 
    .merge(orders, on='customer_id', how='left')
    .merge(order_items, on='order_id', how='left')
)
total_product_purchased = (
    merged 
    .groupby(['customer_id', 'customer_name'])['quantity']
    .sum()
).reset_index(name='total_product_purchased')

total_product_purchased['total_product_purchased'] = (
    total_product_purchased['total_product_purchased'].fillna(0)
) # this is default so not needed
total_product_purchased.head(4)
```

```python
%%sql 
SELECT c.customer_id, c.customer_name, COALESCE(SUM(oi.quantity),0) AS total_product_purchased
FROM retail_analysis.customers c 
LEFT JOIN retail_analysis.orders o 
ON c.customer_id = o.customer_id 
LEFT JOIN retail_analysis.order_items oi 
ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.customer_name;
```

<pre>
Question 37 — Mixed JOIN + Aggregation

Find all customers and their total revenue. 
Include customers who have never placed an order, 
and show 0 revenue for them. 
Sort by total revenue from highest to lowest.

customer_id
customer_name
total_revenue
</pre>

<pre>
DISTINCT customers[customer_name],
(products[selling_price] * order_items[quantity]).sum() AS total_revenue
ORDER BY total_revenue DESC
customers 1* 2* (o revenue for 2*)
products 1* 3
</pre>

```python
merged = (
    customers
    .merge(orders, on='customer_id')
    .merge(order_items, on='order_id')
    .merge(products, on='product_id')
)
merged['total_revenue'] = (
    merged['selling_price'] * merged['quantity']
)
total_revenue = (
    merged
    .groupby(['customer_name', 'customer_id'])['total_revenue']
    .sum()
).reset_index(name='total_revenue')
total_revenue.sort_values('total_revenue', ascending=False).head(4) 
```

```python
%%sql 
SELECT c.customer_name, COALESCE(SUM(p.selling_price * oi.quantity), 0) AS total_revenue
FROM retail_analysis.customers c 
LEFT JOIN retail_analysis.orders o 
ON c.customer_id = o.customer_id
LEFT JOIN retail_analysis.order_items oi 
ON o.order_id = oi.order_id
LEFT JOIN retail_analysis.products p 
ON oi.product_id = p.product_id
GROUP BY c.customer_name,  c.customer_id
ORDER BY total_revenue DESC;
```
